```python
import numpy as np
import pandas as pd

cities = [
    'Tel Aviv', 'Jerusalem', 'London', 'Paris', 'New York',
    'Tokyo', 'Nairobi', 'Dubai', 'Berlin', 'Toronto'
]

months = [
    'January', 'February', 'March', 'April', 'May', 'June',
    'July', 'August', 'September', 'October', 'November', 'December'
]

# Create a sample DataFrame using the lists to define 'df'
df = pd.DataFrame({
    'City': np.random.choice(cities, 12),
    'Month': months,
    'Value': np.random.randint(10, 100, 12)
})

print(df)
```

```python
# Display the first five cities
display(df.head())

# Check dataset dimensions
print("Dataset shape:", df.shape)

# Check minimum and maximum temperatures
print("Minimum temperature:", df['Value'].min(), "°C")
print("Maximum temperature:", df['Value'].max(), "°C")
```

```python
# Calculate the average temperature across all 12 months
annual_avg = df['Value'].mean()

# Display the annual average temperature
print("Annual Average Temperature (°C):")
print(round(annual_avg, 2))
```

```python
# Group by city and calculate the mean for each city
city_averages = df.groupby('City')['Value'].mean()

highest_temp_city = city_averages.idxmax()
highest_avg_temp = city_averages.max()

print("City with the highest average temperature:", highest_temp_city)
print("Highest annual average temperature:", round(highest_avg_temp, 2), "°C")
```

```python
lowest_temp_city = city_averages.idxmin()
lowest_avg_temp = city_averages.min()

print("City with the lowest average temperature:", lowest_temp_city)
print("Lowest annual average temperature:", round(lowest_avg_temp, 2), "°C")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(14, 7))

sns.lineplot(
    data=df,
    x='Month',
    y='Value',
    hue='City',
    palette='tab10',
    marker='o'
)

plt.title('Monthly Temperature Trends Across Cities', fontsize=16)
plt.xlabel('Month')
plt.ylabel('Temperature (°C)')
plt.legend(title='City', bbox_to_anchor=(1.05, 1))
plt.grid(True, alpha=0.3)
plt.tight_layout()

plt.show()
```

I analysis demonstrates how NumPy and Pandas can be used to generate and summarize daily data, and how Matplotlib and Seaborn can visualize patterns and comparisons. The notebook also identifies the cities with the highest and lowest average temperatures using idxmax() and idxmin().

Important: Because the data is randomly generated, rerunning the notebook with a different random seed or different temperature ranges may change the results
